```python
import torch
import torch.nn as nn

class SimpleTransformer(nn.Module):
    def __init__(self, vocab_size=1000, d_model=512):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, d_model)
        self.linear = nn.Linear(d_model, vocab_size)

    def forward(self, src, tgt):
        x = self.embedding(tgt)
        return self.linear(x)
```

```python
import math

class NoamLR:
    def __init__(self, optimizer, d_model, warmup_steps=4000):
        self.optimizer = optimizer
        self.d_model = d_model
        self.warmup_steps = warmup_steps
        self.step_num = 0

    def step(self):
        self.step_num += 1

        lr = (self.d_model ** -0.5) * min(
            self.step_num ** -0.5,
            self.step_num * (self.warmup_steps ** -1.5)
        )

        for g in self.optimizer.param_groups:
            g["lr"] = lr

        return lr
```

```python
def init_wandb(run_name):
    wandb.init(
        project="transformer-assignment3",
        name=noam-vs-fixed
    )
```

```python
import torch.nn.functional as F

def train_one_epoch(model, loader, optimizer, scheduler=None):
    model.train()
    total_loss = 0

    for src, tgt in loader:
        src, tgt = src.to(device), tgt.to(device)

        optimizer.zero_grad()

        output = model(src, tgt[:, :-1])

        loss = F.cross_entropy(
            output.reshape(-1, output.shape[-1]),
            tgt[:, 1:].reshape(-1),
            ignore_index=0
        )

        loss.backward()
        optimizer.step()

        if scheduler:
            lr = scheduler.step()
            wandb.log({"lr": lr})

        wandb.log({"train_loss": loss.item()})

        total_loss += loss.item()

    return total_loss / len(loader)
```

```python
def evaluate(model, loader):
    model.eval()
    correct, total = 0, 0

    with torch.no_grad():
        for src, tgt in loader:
            src, tgt = src.to(device), tgt.to(device)

            output = model(src, tgt[:, :-1])
            preds = output.argmax(-1)

            correct += (preds == tgt[:, 1:]).sum().item()
            total += tgt[:, 1:].numel()

    acc = correct / total
    wandb.log({"val_accuracy": acc})
    return acc
```

```python
def run_noam(model, train_loader, val_loader, d_model, epochs=10):

    init_wandb("Noam_Scheduler")

    optimizer = torch.optim.Adam(model.parameters(), lr=0)
    scheduler = NoamLR(optimizer, d_model=d_model)

    model.to(device)

    for epoch in range(epochs):

        train_loss = train_one_epoch(model, train_loader, optimizer, scheduler)
        val_acc = evaluate(model, val_loader)

        wandb.log({
            "epoch": epoch,
            "train_loss_epoch": train_loss,
            "val_accuracy_epoch": val_acc
        })
```

```python
def run_fixed_lr(model, train_loader, val_loader, lr=1e-4, epochs=10):

    init_wandb("Fixed_LR")

    optimizer = torch.optim.Adam(model.parameters(), lr=lr)

    model.to(device)

    for epoch in range(epochs):

        train_loss = train_one_epoch(model, train_loader, optimizer, scheduler=None)
        val_acc = evaluate(model, val_loader)

        wandb.log({
            "epoch": epoch,
            "train_loss_epoch": train_loss,
            "val_accuracy_epoch": val_acc
        })
```

```python
# ============================================================
# Transformer Training Comparison
# ============================================================

import math
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import wandb

# ============================================================
# CONFIG
# ============================================================

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

BATCH_SIZE = 32
EPOCHS = 10
D_MODEL = 512
WARMUP_STEPS = 4000
FIXED_LR = 1e-4

VOCAB_SIZE = 1000
NUM_CLASSES = 10

PROJECT_NAME = "transformer-assignment3"

global_step = 0

# ============================================================
# MODEL
# ============================================================

class SimpleTransformer(nn.Module):
    def __init__(self, d_model, vocab_size=VOCAB_SIZE, num_classes=NUM_CLASSES, use_scaling=False):
        super().__init__()
        self.use_scaling = use_scaling
        self.d_model = d_model

        self.embedding = nn.Embedding(vocab_size, d_model)

        self.transformer = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(d_model=d_model, nhead=8),
            num_layers=2
        )

        self.fc = nn.Linear(d_model, num_classes)

    def forward(self, x):
        x = self.embedding(x)

        if self.use_scaling:
            x = x * (self.d_model ** 0.5)

        x = self.transformer(x)

        x = x.mean(dim=1)

        x = self.fc(x)

        return x


# ============================================================
# NOAM SCHEDULER
# ============================================================

class NoamScheduler:
    def __init__(self, optimizer, d_model, warmup_steps):
        self.optimizer = optimizer
        self.d_model = d_model
        self.warmup_steps = warmup_steps
        self.step_num = 0

    def step(self):

        self.step_num += 1

        lr = (
            self.d_model ** (-0.5)
            * min(
                self.step_num ** (-0.5),
                self.step_num * (self.warmup_steps ** (-1.5))
            )
        )

        for g in self.optimizer.param_groups:
            g["lr"] = lr

        self.optimizer.step()

        return lr


# ============================================================
# TRAINING
# ============================================================

def train_one_epoch(model, loader, criterion, optimizer, scheduler=None):

    global global_step

    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for batch_x, batch_y in loader:

        batch_x = batch_x.to(DEVICE)
        batch_y = batch_y.to(DEVICE)

        optimizer.zero_grad()

        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)

        loss.backward()

        # ========================================================
        # GRADIENT NORMS
        # ========================================================

        query_grad_norm = 0.0
        key_grad_norm = 0.0

        for name, param in model.named_parameters():

            if param.grad is None:
                continue

            grad_norm = param.grad.data.norm().item()

            name_lower = name.lower()

            # Approximate separation
            if "q" in name_lower:
                query_grad_norm += grad_norm

            if "k" in name_lower:
                key_grad_norm += grad_norm

        # ========================================================
        # OPTIMIZER STEP
        # ========================================================

        optimizer.step()

        if scheduler:
            lr = scheduler.step()
        else:
            lr = optimizer.param_groups[0]["lr"]

        # ========================================================
        # W&B LOGGING (first 1000 steps only)
        # ========================================================

        if global_step <= 1000 and global_step % 5 == 0:
            wandb.log({
                "step": global_step,
                "train_loss": loss.item(),
                "query_grad_norm": query_grad_norm,
                "key_grad_norm": key_grad_norm,
                "learning_rate": lr
            })

        total_loss += loss.item()

        preds = outputs.argmax(dim=1)
        correct += (preds == batch_y).sum().item()
        total += batch_y.size(0)

        global_step += 1

    return total_loss / len(loader), correct / total, lr


# ============================================================
# VALIDATION
# ============================================================

def validate(model, loader, criterion):

    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():

        for batch_x, batch_y in loader:

            batch_x = batch_x.to(DEVICE)
            batch_y = batch_y.to(DEVICE)

            outputs = model(batch_x)
            loss = criterion(outputs, batch_y)

            total_loss += loss.item()

            preds = outputs.argmax(dim=1)
            correct += (preds == batch_y).sum().item()
            total += batch_y.size(0)

    return total_loss / len(loader), correct / total


# ============================================================
# EXPERIMENT RUNNER
# ============================================================

def run_experiment(experiment_name, use_noam=False):

    global global_step
    global_step = 0

    wandb.init(
        project=PROJECT_NAME,
        name=experiment_name,
        reinit=True
    )

    model = SimpleTransformer(D_MODEL).to(DEVICE)

    criterion = nn.CrossEntropyLoss()

    optimizer = optim.Adam(
        model.parameters(),
        lr=FIXED_LR,
        betas=(0.9, 0.98),
        eps=1e-9
    )

    scheduler = None
    if use_noam:
        scheduler = NoamScheduler(optimizer, D_MODEL, WARMUP_STEPS)

    # dummy data
    x_train = torch.randint(0, VOCAB_SIZE, (1000, 20))
    y_train = torch.randint(0, NUM_CLASSES, (1000,))

    x_val = torch.randint(0, VOCAB_SIZE, (200, 20))
    y_val = torch.randint(0, NUM_CLASSES, (200,))

    train_loader = DataLoader(list(zip(x_train, y_train)), batch_size=BATCH_SIZE, shuffle=True)
    val_loader = DataLoader(list(zip(x_val, y_val)), batch_size=BATCH_SIZE)

    for epoch in range(EPOCHS):

        train_loss, train_acc, lr = train_one_epoch(
            model, train_loader, criterion, optimizer, scheduler
        )

        val_loss, val_acc = validate(model, val_loader, criterion)

        print(f"Epoch {epoch+1} | Train Loss: {train_loss:.4f} | Val Acc: {val_acc:.4f}")

        wandb.log({
            "epoch": epoch + 1,
            "train_loss": train_loss,
            "train_accuracy": train_acc,
            "val_loss": val_loss,
            "val_accuracy": val_acc,
            "learning_rate": lr
        })

    wandb.finish()


# ============================================================
# RUN EXPERIMENTS
# ============================================================

run_experiment("Fixed_LR", use_noam=False)
run_experiment("Noam_Scheduler", use_noam=True)
```

```python
# ============================================================
# MODIFY MultiHeadAttention CLASS
# ============================================================

class MultiHeadAttention(nn.Module):

    def __init__(self, d_model, num_heads, use_scaling=True):

        super().__init__()

        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads
        self.use_scaling = use_scaling

        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)

        self.fc_out = nn.Linear(d_model, d_model)

    def split_heads(self, x, batch_size):

        x = x.view(batch_size, -1, self.num_heads, self.d_k)

        return x.transpose(1, 2)

    def forward(self, query, key, value, mask=None):

        batch_size = query.shape[0]

        Q = self.W_q(query)
        K = self.W_k(key)
        V = self.W_v(value)

        Q = self.split_heads(Q, batch_size)
        K = self.split_heads(K, batch_size)
        V = self.split_heads(V, batch_size)

        # ====================================================
        # WITH / WITHOUT SCALING
        # ====================================================

        if self.use_scaling:
            energy = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)
        else:
            energy = torch.matmul(Q, K.transpose(-2, -1))

        if mask is not None:
            energy = energy.masked_fill(mask == 0, float("-1e20"))

        attention = torch.softmax(energy, dim=-1)

        out = torch.matmul(attention, V)

        out = out.transpose(1, 2).contiguous()

        out = out.view(batch_size, -1, self.d_model)

        out = self.fc_out(out)

        return out
```

```python
global_step = 0

def train_one_epoch(model, loader, criterion, optimizer, scheduler=None):

    global global_step

    model.train()

    total_loss = 0
    correct = 0
    total = 0

    for batch_x, batch_y in loader:

        batch_x = batch_x.to(DEVICE)
        batch_y = batch_y.to(DEVICE)

        optimizer.zero_grad()

        outputs = model(batch_x)
        loss = criterion(outputs, batch_y)

        loss.backward()

        query_sq_norm = 0.0
        key_sq_norm = 0.0

        for name, param in model.named_parameters():

            if param.grad is None:
                continue

            grad_norm_sq = param.grad.data.pow(2).sum().item()

            if "q" in name.lower():
                query_sq_norm += grad_norm_sq

            if "k" in name.lower():
                key_sq_norm += grad_norm_sq

        query_grad_norm = query_sq_norm ** 0.5
        key_grad_norm = key_sq_norm ** 0.5

        optimizer.step()
        if scheduler:
            scheduler.step()

        lr = optimizer.param_groups[0]["lr"]

        # IMPORTANT: consistent logging
        if global_step <= 1000 and global_step % 5 == 0:
            wandb.log({
                "step": global_step,
                "train_loss": loss.item(),
                "query_grad_norm": query_grad_norm,
                "key_grad_norm": key_grad_norm,
                "learning_rate": lr
            })

        total_loss += loss.item()

        preds = outputs.argmax(dim=1)
        correct += (preds == batch_y).sum().item()
        total += batch_y.size(0)

        global_step += 1

    return total_loss / len(loader), correct / total, lr
```

```python
def run_experiment(name, use_scaling):

    global global_step
    global_step = 0

    wandb.init(
        project="transformer-assignment3",
        name=name,
        reinit=True
    )

    model = SimpleTransformer(
        d_model=D_MODEL,
        use_scaling=use_scaling
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

    # dummy data
    x = torch.randint(0, VOCAB_SIZE, (1000, 20))
    y = torch.randint(0, NUM_CLASSES, (1000,))

    loader = DataLoader(list(zip(x, y)), batch_size=32, shuffle=True)

    for epoch in range(3):
        train_one_epoch(model, loader, criterion, optimizer)

    wandb.finish()
```

```python
class MultiHeadAttention(nn.Module):

    def __init__(self, d_model, num_heads, use_scaling=True):
        super().__init__()

        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads
        self.use_scaling = use_scaling

        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)

        self.fc_out = nn.Linear(d_model, d_model)

    def split_heads(self, x, batch_size):
        x = x.view(batch_size, -1, self.num_heads, self.d_k)
        return x.transpose(1, 2)

    def forward(self, query, key, value, mask=None):

        B = query.shape[0]

        Q = self.W_q(query)
        K = self.W_k(key)
        V = self.W_v(value)

        Q = self.split_heads(Q, B)
        K = self.split_heads(K, B)
        V = self.split_heads(V, B)

        scores = torch.matmul(Q, K.transpose(-2, -1))

        # ======================================================
        # ABRATION: SCALING FACTOR √(1/dk)
        # ======================================================
        if self.use_scaling:
            scores = scores / math.sqrt(self.d_k)

        if mask is not None:
            scores = scores.masked_fill(mask == 0, float("-1e9"))

        attention = torch.softmax(scores, dim=-1)

        out = torch.matmul(attention, V)

        out = out.transpose(1, 2).contiguous()
        out = out.view(B, -1, self.d_model)

        return self.fc_out(out)
```
